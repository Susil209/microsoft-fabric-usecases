```python
# Read the CSV file
df = spark.read.format("csv").option("header", "true").load("Files/bronze/diabetic_data.csv")

# Replace spaces and invalid characters in column names with underscores
for col_name in df.columns:
    new_col_name = col_name.replace(" ", "_").replace("(", "").replace(")", "").replace(".", "")
    df = df.withColumnRenamed(col_name, new_col_name)

# Save the cleaned dataframe as a Delta table
df.write.format("delta").mode("overwrite").saveAsTable("dbo.bronze_encounters")
```

```python
# Load the saved table to verify what was actually written
saved_df = spark.read.table("dbo.bronze_encounters")

# Get row and column counts
row_count = saved_df.count()
col_count = len(saved_df.columns)

print(f"Row count: {row_count}")
print(f"Column count: {col_count}")
```
